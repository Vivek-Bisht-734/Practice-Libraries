```python
# os.getcwd() - to get current working directory
print(os.getcwd())

print('--' * 30)

# os.chdir(path) - changed cwd to a new location
path = 'C:\\Users\\bisht\\Downloads'
os.chdir(path)
print(os.getcwd())
os.chdir('..') # Go up one folder (parent folder)
print(os.getcwd())
os.chdir('C:\\Users\\bisht\\Documents')
print(os.getcwd())


print('--' * 30)

# os.listdir(path = '.')  --> here '.' always means current dir, returns list of all file and folders 
# inside a given dir
os.chdir('C:\\Users\\bisht\\jupyter_env\\python_libs')
print(os.listdir())   # this also does same :   os.listdir(path = '.')
print(os.listdir("C:\\Users\\bisht\\jupyter_env"))

for item in os.listdir(path = '.'):
    print(item)

print('\n\n', '=' * 80)

# os.scandir(path = '.') == Smarter version of os.listdir(), nor just names it returns specil DirEntry
# objects that contain extra info about each file/folder (like file or a folder and when it was last
# modified wihtout extra function calls)
path = 'C:\\Users\\bisht\\jupyter_env\\python_libs'
with os.scandir(path) as entries:
    for entry in entries:
        if entry.is_file():
            print(f'FILE: {entry.name} ({entry.path})')
        if entry.is_dir():
            print(f'FOLDER: {entry.name}')
            print(f'File Size: {entry.stat().st_size} bytes')
        if entry.is_symlink():
            target_path = os.readlink(entry.path)
            print(f'SYMLINK: {entry.name} -> points to: {target_path}')

# *** Use os.scandir() instead of os.listdir() when we need to know if something is file nor folder.
# it is faster because it avoids extra lookups.***
```

```python
# os.path - contains collection of fns designed to work with file paths. Auto handles difference b/w
# windows backslashes and linux/max forward slashes

# 1. os.path.join(*paths) -->   os.path.join(path1, path2, path3, .....)
path = os.path.join('C:\\', 'Users', 'bisht', 'Videos')
print(path)

# If a component is an absolute path, it RESETS and starts fresh
result = os.path.join('/home/user', '/etc/config')
print(result)

print('--' * 30, '\n')

# 2. os.path.exists(path) --> checks whether a file or folder actually exists at the given path.
print(os.path.exists("C:\\Users\\bisht\\Videos"))
print(os.path.exists("C:\\Users\\bisht\\Videos\\hehe.mp3"))

print('--' * 30, '\n')

# 3. os.path.isfile(path) --> checks if path points to a file and 
# os.path.isdir(path) --> checks if it points to a folder.
print(os.path.isfile("C:\\Users\\bisht\\Videos\\hehe.mp3"))
print(os.path.isfile("C:\\Users\\bisht\\Downloads\\stats.txt"))
print(os.path.isdir("C:\\Users\\bisht\\Downloads\\illapode"))
print(os.path.isdir("C:\\Users\\bisht\\Downloads\\Pantech"))


print('--' * 30, '\n')

# 4. os.path.split(path) --> breaks a path into two parts: the folder and the filename part
# os.path.basename(path) / os.path.dirname(path) --> gives just the filename. dirname() gives just the 
# folder part
path = "C:\\Users\\bisht\\jupyter_env\\python_libs\\OS_and_SYS\\os_and_sys.ipynb"
folder, filename = os.path.split(path)
print(filename)
print(folder)

print(os.path.basename(path))
print(os.path.dirname(path))


print('--' * 30, '\n')

# 5. os.path.splitext(path) --> separates filename from its extension (the part after last dot, like .py)
path = "C:\\Users\\bisht\\jupyter_env\\python_libs\\OS_and_SYS\\os_and_sys.ipynb"
path2 = "C:\\Users\\bisht\\jupyter_env\\python_libs\\OS_and_SYS"
name, ext = os.path.splitext(path)
print(name)
print(ext)
name, ext = os.path.splitext(path2)
print(name)
print(ext)

print('--' * 30, '\n')

# 6. os.path.abspath(path) --> converts relative path into an absolute path based on cwd. It also
# cleans up .. or . components
print(os.path.abspath('Documents'))
print(os.path.abspath('.'))
print(os.path.abspath('..'))
print(os.path.abspath('../'))
print(os.path.abspath('../Zoom'))

print('--' * 30, '\n')

# 7. os.path.getsize(path) - returns size of file in bytes
path = "C:\\Users\\bisht\\jupyter_env\\python_libs\\OS_and_SYS\\os_and_sys.ipynb"
size = os.path.getsize(path)
print(size, 'bytes')
print(f'Size in KB: {size / 1024}')
print(f'Size in MB: {size / (1024 * 1024)}')

print('--' * 30, '\n')

# 8. os.path.getmtime(), os.path.getatime(), os.path.getctime() --> return timestamps for when a file
# was last modified, accessed and created. The timestamps are in format called Unix timestamp.
import datetime

path = "C:\\Users\\bisht\\jupyter_env\\python_libs\\OS_and_SYS\\os_and_sys.ipynb"
mtime = os.path.getmtime(path)  # modified time
print(mtime)
readable = datetime.datetime.fromtimestamp(mtime)
print(readable)

atime = os.path.getatime(path)  # accessed time
print(atime)
readable = datetime.datetime.fromtimestamp(atime)
print(readable)

ctime = os.path.getctime(path) # created time
print(ctime)
readable = datetime.datetime.fromtimestamp(ctime)
print(readable)
```

```python
# Creating and Removing Directories
# os.mkdir(path, mode = 0o777) --> creates single new dir
os.chdir("C:\\Users\\bisht\\jupyter_env\\python_libs\\OS_and_SYS")
print(os.getcwd())

path = os.getcwd()
# os.mkdir('MKDIR_TEST_1', mode = 0o777)  # 0o777 means everyone can read / write and execute it

# Another safe way to create folder
if not os.path.exists('MKDIR_TEST_1'):
    os.mkdir('MKDIR_TEST_1')
    print('Folder Created')
else:
    print('Folder already exists')

print('--' * 30, '\n')

# os.makedirs(path, mode = 0o777, exist_ok = False) --> creates a dir and all missing parent dirs.
os.makedirs('MKDIRS_TEST_2/Second_File/Third_File', mode = 0o777, exist_ok = True)

print('--' * 30, '\n')

# os.rmdir(path) --> removes single, empty dir. Will fail if dir is not empty -- we must del all files
# inside the dir first.
os.rmdir('MKDIR_TEST_1')

print('--' * 30, '\n')

# os.removedirs(path) --> removes empty dirs and keeps going up to remove empty parent dirs too. Stops
# when it hits a non-empty dir.
os.removedirs('MKDIRS_TEST_2/Second_File/Third_File')

print('--' * 30, '\n')
```

```python
# Environment Variables: It stores settings like home dir, system language, python path and other configs

# os.environ - it is dict like object that holds all current environment variables.
for key, value in os.environ.items():
    print(f'{key} = {value}')  # prints all environment variables

print('--' * 40)

home = os.environ['HOMEDRIVE'] # Read specific variables
print(home)

print('--' * 40)

path = os.environ['PATH']
print(path)

print('---' * 40)

# os.environ.get(key, default = None) --> safely reads an environment variable. If variable doesn't 
# exist, it returns default value instead of crashing.
debug = os.environ.get('DEBUG_MODE', 'Try 1')
print(debug)

debug1 = os.environ.get('DEBUG_MODE', 'Try 2')
print(debug1)

# common pattern for reading configs
database_url = os.environ.get('DATABASE_URL', 'sqlite:///local.db')
port = int(os.environ.get('PORT', 8000))
debug = os.environ.get('DEBUG', 'false').lower() == 'true'
print(debug)

print('---' * 40)

# os.getenv(key, default = None) --> same like os.environ.get() -- shorter way to write same thing.
home = os.environ.get('HOMEDRIVE', '/tmp')
home = os.getenv('HOMEDRIVE', '/tmp')
print(home)

print('---' * 40)

# os.putenv(key, value) and setting via os.environ --> sets or updates an environment variable.
# Best way to update is os.environ directly like a dict, because os.putenv() alone does not update
# the os.environ

# Recommended Way
os.environ['MY_VAR'] = 'Hello_World'
print(os.environ.get('MY_VAR'))

# Add multiple new variable
os.environ['APP_NAME'] = 'MyApp'
os.environ['APP_VERSION'] = '1.0.0'
os.environ['LOG_LEVEL'] = 'INFO'
# changes to os.environ are temporary only for current python process and its children. When program exits
# the changes are gone. Our system is not permanently modified.

print('---' * 40)

# os.unsetenv(key) and Deleting from os.environ --> removes environment variable. The recommmended way
# is to use: "del os.environ[key]" or "os.environ.pop[key"
os.environ['TEMP_VAR'] = 'temporary'

del os.environ['TEMP_VAR'] # deleted it

# or POP it (also returns the values before deleting)
value = os.environ.pop('TEMP_VAR', None) # None = default if not found
print(os.environ.get['TEMP_VAR'])
```
